#### subplots and legends

```python
days=[1,2,3,4,5]
original_price=[20,35,46,50,68]
predicted_price=[18,33,40,44,55]
```

```python
plt.plot(days,original_price,'b',label="original_price")
plt.plot(days,predicted_price,'r',label="predicted_price")
plt.legend()
plt.title("plot with legends")
plt.xlabel("days")
plt.ylabel("prices")
```

#### subplots

```python
#first subplot- oiginal price
plt.subplot(3,1,1)
plt.plot(days,original_price,'b-o',label="original_price")
plt.title("Original Price over time")
plt.xlabel("weekdays")
plt.ylabel("prices")
plt.legend()

#seconda subplot- predicted price
plt.subplot(3,1,2)
plt.plot(days,predicted_price,'r-o',label="predicted_price")
plt.title("Predicted Price over time")
plt.xlabel("weekdays")
plt.ylabel("prices")
plt.legend()

#THIRD subplot- predicted price
price_diff=[o-p for o, p in zip(original_price,predicted_price)]
plt.subplot(3,1,3)

plt.bar(days,price_diff,color='orange',label="Price difference")
plt.title("Difference between original Price and Predicted Price")
plt.xlabel("Days")
plt.ylabel("Price Difference")
plt.legend()

plt.tight_layout()
         
```

### saving the figure

```python
plt.plot(days,original_price,'b',label="original_price")
plt.plot(days,predicted_price,'r',label="predicted_price")
plt.legend(loc=4)
plt.title("plot with legends")
plt.xlabel("days")
plt.ylabel("prices")

plt.savefig("exmple_predction.png",dpi=400)
```
